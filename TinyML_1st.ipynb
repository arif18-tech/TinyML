````python
import math
import matplotlib.pyplot as plt
````

````python
w=3
b=0
x=[-1,0,1,2,3,4]
y=[-3,-1,1,3,5,7]
myY=[]
for thisX in x:
  thisY=(w*thisX)+b
  myY.append(thisY)

print("real Yis " + str(y))
print("My Y is " + str(myY))
````

````python
#visualize loss/errors
plt.scatter(x, y, color='blue', label='Actual (y)')
#ground truth
plt.scatter(x, myY, color='green', label='Predicted (myY)')
# predicted errors
plt.vlines(x, ymin=min(y,myY), ymax=max(y,myY), color="red", linestyle="--", alpha=0.5, label='Errors')
plt.legend()
plt.show()
````

````python
#let's calculate the loss
total_square_error = 0
size = len(y)
for i in range(0, size):
  square_error = (y[i] - myY[i]) ** 2
  total_square_error += square_error
total_square_error /= size
print("My loss is:" + str(math.sqrt(total_square_error)))
````

### **Line 4**

This is probably the most confusing line. Let's understand it slowly.

First, remember:

\\[ y=wx+b \\]

We have two parameters:

- \\(w\\)
- \\(b\\)

We want to know how changing each parameter affects our loss.

Mathematically, we calculate:

\\[ dw=\frac{\partial Loss}{\partial w} \\]

\\[ db=\frac{\partial Loss}{\partial b} \\]

What do these mean?

- `dw` tells us how the loss changes when we change \\(w\\).
- `db` tells us how the loss changes when we change \\(b\\).

Now look at:

```
t.gradient(current_loss, [model.w, model.b])
```

We are asking TensorFlow:

**"Please calculate the derivatives of the loss with respect to w and b."**

TensorFlow calculates both derivatives and stores them in:

```
dw, db
```

iframe

Think of these derivatives as directions that tell us how to adjust our parameters to reduce the error.

### **Line 5**

```

```

Now we know how changing \\(w\\) affects our loss.

So, we update \\(w\\).

The mathematical equation is:

\\[ w\_{\text{new}}=w\_{\text{old}}-\alpha dw \\]

Where:

- \\(w\_{\text{old}}\\) = Previous value of w
- \\(\alpha\\) = Learning rate
- \\(dw\\) = Derivative of loss with respect to w
- \\(w\_{\text{new}}\\) = Updated value of w

What does `assign_sub()` mean?

It means:

**Subtract the calculated value from the existing value of w and store the result back in w.**

For example:

Suppose:

\\[ w=10 \\]

\\[ learning\\\_rate=0.1 \\]

\\[ dw=2 \\]

Then:

\\[ w\_{\text{new}}=10-(0.1\times2) \\]

\\[ \boxed{w\_{\text{new}}=9.8} \\]

So, our old w was 10, and now it becomes 9.8.

Mathematically:

\\[ b\_{\text{new}}=b\_{\text{old}}-\alpha db \\]

We update the value of \\(b\\) to help reduce the loss.

These three steps are the foundation of **Gradient Descent**.

One small technical correction to your code: the parameter updates should normally happen outside the `with tf.GradientTape()` block. A cleaner version is:

```
def train(model, inputs, outputs, learning_rate):    with tf.GradientTape() as t:        current_loss = loss(model(inputs), outputs)    dw, db = t.gradient(current_loss, [model.w, model.b])    model.w.assign_sub(learning_rate * dw)    model.b.assign_sub(learning_rate * db)    return current_loss
```

````python
#Define our training procedure
def train(model, inputs, outputs, learning_rate):
  with tf.GradientTape() as t:
    #loss function
    current_loss = loss(model(inputs), outputs)
    #Here is where you differentiate loss function w.r.t model parameters
    dw, db = t.gradient(current_loss, [model.w, model.b]) #dloss/dw, dloss/db

    # And here is where you update the model parameters based on the learning rate chosen
    model.w.assign_sub(learning_rate * dw) #model.w model.w learning_rate dw
    model.b.assign_sub(learning_rate * db) #model.b model.b learning rate db
    return current_loss
````

## 1. `class` ki?

Class holo ekta **nokshar naksha (blueprint)**. Dhoro, "Gari" er naksha ache. Oi naksha theke tumi onek gari banate paro, prottekta gari-r nijer rong, nijer speed thake.

```python
class Dog:
    pass

d1 = Dog()   # naksha theke ekta kukur banalam
d2 = Dog()   # arekta kukur
```

`d1`, `d2` ke bole **object** (ba instance).

## 2. `__init__` ki?

Jokhon tumi object banao (`Dog()`), tokhon Python **auto-matic** `__init__` function-ta chalay. Eta hocche **setup korar jayga**. Notun object-er jonno shuru-r jinis gulo ekhane set kora hoy.

(Note: tumi `__int__` likhecho, kintu ashol nam `__init__`. `init` = initialize.)

```python
class Dog:
    def __init__(self, name):
        self.name = name

d1 = Dog("Tommy")
print(d1.name)   # Tommy
```

## 3. `self` ki?

`self` mane **"ei object-ta nijei"**. Jokhon `d1 = Dog("Tommy")` likho, tokhon `self` = `d1`. Tai `self.name = name` mane "ei kukur-er name rakho Tommy".

`self.name` likhle data object-er bhitore **save** thake, pore onno function theke abar pawa jay.

## 4. `__call__` ki?

Normally object ke function-er moto chalano jay na. Kintu `__call__` thakle **object-ke function-er moto call kora jay**.

```python
class Double:
    def __call__(self, x):
        return x * 2

d = Double()
print(d(5))   # 10   <-- object-ke function-er moto chalalam
```

`d(5)` likhle Python mone mone `d.__call__(5)` chalay.

class Model(object):
